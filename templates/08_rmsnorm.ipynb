````python
def rms_norm(x, weight, eps=1e-6):
    mean_square = x.pow(2).mean(dim=-1, keepdim=True)  # Compute mean square, shape (batch_size, seq_length, 1)
    rms = torch.sqrt(mean_square + eps)  # Compute root mean square, shape (batch_size, seq_length, 1)
    x_normalized = x / rms  # Normalize input, shape (batch_size, seq_length, hidden_size)
    output = x_normalized * weight  # Scale by weight, shape (batch_size, seq_length, hidden_size)
    return output
````

````python
# 🧪 Debug
x = torch.randn(2, 8)
w = torch.ones(8)
out = rms_norm(x, w)
print("Output shape:", out.shape)
print("RMS of output:", out.pow(2).mean(dim=-1).sqrt())  # should be ~1
````
